```python
with Session() as session:
    category_names = [
        "Breakfast",
        "Lunch",
        "Dinner",
        "Dessert",
        "Vegan",
        "Italian",
        "Gluten-Free",
        "Quick Meals",
    ]

    ingredient_names = [
        "Flour",
        "Milk",
        "Egg",
        "Sugar",
        "Butter",
        "Pasta",
        "Tomato",
        "Basil",
        "Garlic",
        "Olive Oil",
        "Parmesan",
        "Chickpeas",
        "Avocado",
        "Bread",
        "Cocoa",
        "Banana",
        "Oats",
        "Spinach",
        "Chicken Breast",
        "Rice",
    ]

    existing_categories = {category.name for category in session.query(Category).all()}
    existing_ingredients = {ingredient.name for ingredient in session.query(Ingredient).all()}

    for name in category_names:
        if name not in existing_categories:
            session.add(Category(name=name))

    for name in ingredient_names:
        if name not in existing_ingredients:
            session.add(Ingredient(name=name))

    session.commit()

print("Seeded categories and ingredients.")
```

```python
with Session() as session:
    users = {user.email: user for user in session.query(User).all()}
    recipes = {recipe.title: recipe for recipe in session.query(Recipe).all()}

    follows_data = [
        ("ana@foodcraft.local", "marc@foodcraft.local"),
        ("laia@foodcraft.local", "ana@foodcraft.local"),
        ("pau@foodcraft.local", "laia@foodcraft.local"),
        ("marc@foodcraft.local", "ana@foodcraft.local"),
    ]

    favorites_data = [
        ("marc@foodcraft.local", "Classic Pancakes"),
        ("ana@foodcraft.local", "Spaghetti al Pomodoro"),
        ("laia@foodcraft.local", "Avocado Toast Deluxe"),
        ("pau@foodcraft.local", "Chocolate Mug Cake"),
        ("ana@foodcraft.local", "Vegan Buddha Bowl"),
    ]

    ratings_data = [
        ("marc@foodcraft.local", "Classic Pancakes", 5, "Perfect for weekend breakfast."),
        ("ana@foodcraft.local", "Spaghetti al Pomodoro", 4, "Simple and very tasty."),
        ("pau@foodcraft.local", "Chocolate Mug Cake", 5, "Fast and rich dessert."),
        ("laia@foodcraft.local", "Vegan Buddha Bowl", 5, "Healthy and satisfying."),
        ("marc@foodcraft.local", "Chicken Rice Skillet", 4, "Great one-pan dinner."),
    ]

    existing_follows = {
        (follow.follower_id, follow.followed_id)
        for follow in session.query(Follow).all()
    }
    existing_favorites = {
        (favorite.user_id, favorite.recipe_id)
        for favorite in session.query(Favorite).all()
    }
    existing_ratings = {
        (rating.user_id, rating.recipe_id)
        for rating in session.query(Rating).all()
    }

    for follower_email, followed_email in follows_data:
        follower = users[follower_email]
        followed = users[followed_email]
        key = (follower.user_id, followed.user_id)
        if key not in existing_follows:
            session.add(Follow(follower_id=follower.user_id, followed_id=followed.user_id))

    for user_email, recipe_title in favorites_data:
        user = users[user_email]
        recipe = recipes[recipe_title]
        key = (user.user_id, recipe.recipe_id)
        if key not in existing_favorites:
            session.add(Favorite(user_id=user.user_id, recipe_id=recipe.recipe_id))

    for user_email, recipe_title, stars, comment in ratings_data:
        user = users[user_email]
        recipe = recipes[recipe_title]
        key = (user.user_id, recipe.recipe_id)
        if key not in existing_ratings:
            session.add(
                Rating(
                    user_id=user.user_id,
                    recipe_id=recipe.recipe_id,
                    stars=stars,
                    comment=comment,
                )
            )

    session.commit()

print("Seeded follows, favorites and ratings.")
```
